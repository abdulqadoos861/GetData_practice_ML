To forward-fill missing values in a column, you can use the `.ffill()` method. This method propagates the last valid observation forward to next valid observation.

```python
# Forward-fill missing values in the 'age' column
df['age'].ffill(inplace=True)

print('Null values after forward-filling:')
display(df.isnull().sum())
```

#Categorical data

Survived Column Univariate analysis

**a. count plot**

```python
sns.countplot(x="survived",data=df , palette="hls")
```

**b.Pie Chart**

```python
df["survived"].value_counts().plot(kind="pie")
plt.xlabel("Survived")
plt.ylabel("Count")
plt.title("Survived Count")
```
