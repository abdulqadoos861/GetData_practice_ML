```python
from IPython.core import excolors
def rename(name):
  if name == "Royal Challengers Bangalore":
    return "RCB"
  elif name == "Kings XI Punjab":
    return "KXIP"
  elif name == "Delhi Daredevils":
    return "DC"
  elif name == "Deccan Chargers":
    return "SRH"
  elif name == "Rajasthan Royals":
    return "RR"
  elif name == "Mumbai Indians":
    return "MI"
  elif name == "Chennai Super Kings":
    return "CSK"
  elif name == "Kolkata Knight Riders":
    return "KKR"
  elif name == "Sunrisers Hyderabad":
    return "SRH"
  elif name == "Rising Pune Supergiants":
    return "RPS"
  elif name == "Gujarat Lions":
    return "GL"
  elif name == "Delhi Capitals":
    return "DC"
  else:
    return name
```

```python
IPL_df = pd.read_csv('/content/drive/MyDrive/Learning ML/GetData/IPL Matches 2008-2020.csv' , converters={
        "team1": rename,
        "team2": rename,
        "toss_winner": rename,
        "winner": rename
    })
IPL_df
```
