```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
```

```python
df =sns.load_dataset("titanic")
df.head(2)
```

```python
# survived , pclass , sex , age, sibsp,parch, fare, embarked
df.drop(columns=["class","who","adult_male",	"deck",	"embark_town",	"alive","alone"] , inplace=True)
```

```python
df
```

```python
X = df.drop(columns=["survived"])
Y = df["survived"]
```

```python
# Step1 split train test
X_train , X_test , Y_train , Y_test = train_test_split(X , Y, test_size=0.2 , random_state=1)
X_train.head()
```

```python
df.isnull().sum()
```

```python
# Step2 Apply Transofrmation.
si_age = SimpleImputer()
si_embarked = SimpleImputer(strategy="most_frequent")

X_train_age = si_age.fit_transform(X_train[['age']])
X_train_embarked = si_embarked.fit_transform(X_train[['embarked']])

X_test_age = si_age.transform(X_test[['age']])
X_test_embarked = si_embarked.transform(X_test[['embarked']])
#
```

```python
X_train_age
```

```python
# Encoding
# One Hot encoding because these columns are minimal coulmns not ordinal coulmns.
ohe_sex = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
ohe_embarked = OneHotEncoder(sparse_output=False , handle_unknown= "ignore")

X_train_ohe_sex = ohe_sex.fit_transform(X_train[['sex']])
X_train_ohe_embarked = ohe_embarked.fit_transform(X_train_embarked)

X_test_ohe_sex = ohe_sex.transform(X_test[['sex']])
X_test_ohe_embarked = ohe_embarked.transform(X_test_embarked)
```

```python
X_train = X_train.drop(columns = ['sex','age','embarked'])
X_test = X_test.drop(columns = ['sex','age','embarked'])
X_train_transformed = np.concatenate((X_train , X_train_age ,X_train_ohe_sex, X_train_ohe_embarked),axis=1 )
X_test_transformed = np.concatenate((X_test,X_test_age ,X_test_ohe_sex, X_test_ohe_embarked) , axis=1 )
```

```python
#  now training model
clf = DecisionTreeClassifier()
clf.fit(X_train_transformed , Y_train)
y_pred = clf.predict(X_test_transformed)
```

```python
import sklearn.metrics as metrics
print(metrics.accuracy_score(Y_test , y_pred))
```
